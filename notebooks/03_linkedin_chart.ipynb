```python
import matplotlib.pyplot as plt

skills = [
    ("Dashboards / visualization", 83),
    ("SQL", 78),
    ("Power BI", 57),
    ("Excel", 49),
    ("Data cleaning / quality", 47),
    ("Python", 44),
    ("Tableau", 40),
    ("Statistics", 39),
]
n_total = 102

labels = [s for s, _ in skills][::-1]
counts = [c for _, c in skills][::-1]
pcts = [100 * c / n_total for c in counts]

fig, ax = plt.subplots(figsize=(8, 5), dpi=200)
bars = ax.barh(labels, pcts, color="#1f4e79")
for bar, c, p in zip(bars, counts, pcts):
    ax.text(bar.get_width() + 1, bar.get_y() + bar.get_height() / 2,
            f"{p:.0f}%  ({c} of {n_total})", va="center", fontsize=9)

ax.set_xlim(0, 105)
ax.set_xlabel("Share of Data Analyst postings that state the skill (%)")
ax.set_title("Skills stated in 102 Data Analyst job postings in India", loc="left", fontsize=12, fontweight="bold")
for side in ["top", "right"]:
    ax.spines[side].set_visible(False)
fig.text(0.01, 0.01,
         "Source: 300 postings collected by hand from Naukri, LinkedIn, Indeed and Foundit, cleaned to 246; "
         "102 are Data Analyst-titled. Skills read from description text.",
         fontsize=6.5, color="#555555")
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig("../dashboard/linkedin_skills_chart.png")
plt.show()
```
